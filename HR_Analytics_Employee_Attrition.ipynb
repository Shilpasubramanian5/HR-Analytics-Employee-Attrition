```python
attrition_percent = df['Attrition'].value_counts(normalize=True) * 100

plt.figure(figsize=(6,4))

attrition_percent.plot(kind='bar')

plt.title('Employee Attrition Percentage')
plt.xlabel('Attrition')
plt.ylabel('Percentage')

plt.show()
```

```python
department_attrition = pd.crosstab(
    df['Department'],
    df['Attrition'],
    normalize='index'
) * 100

department_attrition
```

```python
department_attrition['Yes'].sort_values(ascending=False)
```

```python
bins = [18, 25, 35, 45, 55, 60]
labels = ['18-25', '26-35', '36-45', '46-55', '56-60']

df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=bins,
    labels=labels
)
```

```python
age_attrition = pd.crosstab(
    df['AgeGroup'],
    df['Attrition'],
    normalize='index'
) * 100

age_attrition
```

```python
age_attrition['Yes'].sort_values(ascending=False)
```

```python
age_attrition['Yes'].plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Attrition Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=0)

plt.show()
```

```python
heatmap_data = satisfaction_involvement['Yes'].unstack()

plt.figure(figsize=(8,5))
sns.heatmap(
    heatmap_data,
    annot=True,
    fmt='.1f',
    cmap='YlOrRd'
)

plt.title('Attrition Rate by Job Satisfaction and Job Involvement')
plt.xlabel('Job Involvement')
plt.ylabel('Job Satisfaction')
plt.show()
```
